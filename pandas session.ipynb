## GROUPBY

```python
# is used to arrange/ group identical data into groups using specified columns e.g if a particular column
# has identical entries in different rows then it will put these rows in one group.
#Group by single column is to place all rows with same value of only that particular column in one group.
# Groupby multiple columns is to place all rows with same value of both columns in one group.eg column 1+column 2 together .
```

```python
#Group by multiple columns means to place all the rows with the same value of both the column1 and column 2 in one group
#Groupby method works with aggregate fuctionss to give us meaningful results,by itself
# it is useless.Example of aggregate fuctions
#sum(): sum of group values is the most important 
#mean()mean of groups
#size():compute group sizes
#count():compute count of groups
#std():standard deviation of groups
#var()variance
#sem() std error of the mean of the groups
#describe()generate descriptive statistics 
#first() first of group values
#last()last of group values
#nth()takes the n value,or a subset if n is a list
#min()mean of group values
#max()max of group values.
```

```python
# we want to group the merged dataframe by counties so we can generate county level aggregates.
```

```python
# combine different  entries based on their county. use census data to aggregate all entries from a certain county together
#say census .groupby(' ')pass the column in which the aggregate is to b performed on.e.g county coz we want
#all the entries with the same county name to be combined. then use sum ()fuction.
census.groupby('County')
census 
```

```python
census.groupby('County')
```

```python
census.groupby('County').sum()  # this will combine different entries under the same county ad sumup the different numerical
                                  #values with the same county
```

```python
# index/ row labels is usually based on the column used to do the groupby.if u want the column to b part of the data frame,
# append the reset_index() to the operation.eg which replace the column index into numerical index.

census.groupby('County').sum().reset_index()

#after you run, the new index/row labels (0,1,2,3) appear
```


### SORTING 
### Use sort_values() method

```python
# if sorting data with numerical columns, u haave to pass in a column name  to the by parameter,otherwise an errorr 
#will be generated.pandas need to know on whichever date/ column to perform the sorting
# to sort your values in a cetain order we use sort_value method . it will rearrange the position of values in 
#descending or ascending order.
```

```python
Census_grouped= census.groupby('County').sum().reset_index()
```

```python
Census_grouped
```

```python
Census_grouped.sort_values(by='Population') # means we are sorting by population. from the lowest populated area to the highest
# (default setting ) sort_value() sort by ascending order.
```

```python
 # want the index values to maintain order as before sorting.
  #  to do that we add reset_index() when we run a new index is included in the left side as follows 

Census_grouped.sort_values(by='Population').reset_index()
```

```python
# the problem now is the old disorganised index is still on the table. to remove it:
# in reset_index(drop=True) pass drop=True.

Census_grouped.sort_values(by='Population').reset_index(drop=True) # sorts in ascending-lowest to highest. to change this,
there argument you pass inside sort_values(ascending=False) ascending and give it the value False
```

```python
# sorts in ascending-lowest to highest. to change this,
# there argument you pass inside sort_values(ascending=False) ascending and give it the value False like these:

Census_grouped.sort_values(by='Population',ascending=False).reset_index(drop=True)
```

```python
# if u want to get the top 10 values/countys with highest population 
# include the head() method. inside the head() you specify the number of top values that you want to get.eg .head(10)

Census_grouped.sort_values(by='Population',ascending=False).reset_index(drop=True).head(10)
```

```python
# if you want to get the bottom 10, use .tail() e,g

Census_grouped.sort_values(by='Population',ascending=False).reset_index(drop=True).tail(10)
```

```python
# the best way to get the bottom values is by using/sorting values in ascending order.then append the head method(),
# then specify the number of values you want.eg the bottom 10  do as follows: .head(10)

Census_grouped.sort_values(by='Population').reset_index(drop=True).head(10)
```
