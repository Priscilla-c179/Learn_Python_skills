````python
import pandas as pd
import numpy as np
import seaborn as sns
````

## Load the dataset

````python
df2=sns.load_dataset("penguins")
````

````python
# get some dataset information
df2.info()
````

The dataset consist of 7 columns and 344 rows.The datatypes  are float for numerical columns amd objects for categorical columns.
There are  some null values in the dataset.

````python
# get the first 5 columns
df2.head()
````

## Play with relational plots

````python
# scatterplot
sns.scatterplot(data=df2,x="flipper_length_mm",y="bill_length_mm",hue="species")
````

````python
#relplot
sns.relplot(data=df2,x="flipper_length_mm",y="body_mass_g",hue="species",kind="scatter")
````

````python
# swarm plot
sns.swarmplot(data=df2,x="species",y="body_mass_g",hue="sex")
````

## Play with distribution plots

````python
sns.histplot(data=df2,x="body_mass_g",hue="species",multiple="stack",kde=True)
````

Gentoo species has the highest body mass,followed by Chinstrap and lastly Adelie.However, the Adelie species has the highest count.

````python
# distplot
sns.displot(data=df2,x="bill_length_mm",hue="species",kind="ecdf")
````

Adelie has the shortest bill of all. her bills start at around 33mm and ends at approximately 45mm.

Gentoo has the longest bill length with a  a range of (44-~60mm). it is skewed towards high numbers/values.there's slight overlap with Chinstrap.

Chinstrap separate Gentoo and Adelie.Has a range of 41-59mm,overlap Gentoo and has longer bills compared to Adelie


## Play with other plots

````python
sns.barplot(data=df2,x="species",y="bill_depth_mm",hue="sex")
````

male has deeper bill compared to female over all. Gentoo has the smallest of all bill depth in terms of both male and female.
