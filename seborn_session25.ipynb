 SEABORN LIBRARY
 
  Seaborn library is a data visualization  tool .it is an upgrade from Matplotlib library . Widely use in data science and machine learning  for exploratory data analysis.
  
 1. It visualizes data distribution: to show how dataset is spread.(Histogram,rug and KDE plots)

 2. Explore relationships between data points or variables.(correlation between 2 features.scatter,line and regression plots.)

 3. Compare categories(compare sales across different quaters).Bar,box,swarm and violin plots.





```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Load the dataset
df=sns.load_dataset('tips')
```

```python
#get the first 5 rows
df.head()
```

```python
# Relational plots
sns.relplot(data=df,x='total_bill',y='tip')
# Relational plots with hue
sns.relplot(data=df,x='total_bill',y='tip',hue='sex')
sns.relplot(data=df,x='total_bill',y='tip',hue='day')
sns.relplot(data=df,x='total_bill',y='tip',hue='time')
sns.relplot(data=df,x='total_bill',y='tip',hue='smoker',col='time')
sns.relplot(data=df,x='total_bill',y='tip',hue='size')
sns.relplot(data=df,x='total_bill',y='tip',hue='size',size='size')
```

```python
sns.relplot(data=df,x='total_bill',y='tip',hue='size',size='size',sizes=(15,200))
```

```python
sns.relplot(data=df,x='total_bill',y='tip',hue='size',alpha=(0.5),size='size')
```

```python
sns.relplot(data=df,x='day', y='total_bill')
```

Based on the above saturday had the highest sales or bill.

```python
# show the relationship between day and total_bill with line
sns.relplot(data=df,x='day', y='total_bill', kind='line')
```

 The line above indicates the mean total bill for each day

```python
sns.relplot(data=df,x='day', y='total_bill', kind='line',hue='sex')
```

From above  the male and female customers behaves differntly on different days based on the total bills.

```python
sns.relplot(data=df,x='day', y='total_bill', kind='line', col='time',hue='sex')
```

Distribution Plots

Distribution plots shows how the continous variable values are spread.

It help i thinking of how often does each value occur in the data.

Distribution plot indicate:

1. shape of the data whether normal,skewed  or uniform

2. Central tendency- Where most values are concentraated(mean,mode, median)

3. Spread or Variability-How wide the data points are spread(standard deviation and variance)

4. SKEWNESS- this is when the data points has a long tail eith on the right skewness or left skewness.

In summary a distribution plot  review whether the data is normal,how values are spread or it contains noise(outliers)


```python
# distplot of tips and bill
sns.displot(data=df,x='total_bill')
```

```python
sns.displot(data=df,x='total_bill',col='sex')
```

```python
# compare using hue
sns.displot(data=df,x='total_bill',hue='sex')
```

```python
# compare with sex and day
sns.displot(data=df,x='total_bill',hue='sex',col='day')
```

```python
sns.displot(data=df,x='total_bill',hue='sex', kind='kde')
```

In this Kernel Density Estimate(KDE)plot,X axis is total bill while Y axis is Density which indicates the probability of bills falling in the range given.

Female bills peak at around 14-17,while male bills peak at 19-20.

The male bill is spread out more compared to female which is narrow and most of it ,< 40 while than of male is ==60.
Both distribution are right skewed indicating there are some high cost bills.

The frequency/Density - the male curve is taller at the peak compared to female indicating that male contributed a larger proportion of the total_bills.

this concludes that men spend more and have the highest bill.

```python
# multiple distribution
sns.displot(data=df,x='total_bill',y='tip', kind='kde', hue='sex')
```

```python
# add rug
sns.displot(data=df,x='total_bill',y='tip', kind='kde', rug=True)
```

rug plot allows us to see the real data points that the KDE is estimating.

It also detects clucter.where many ticks are clustered together that's where the data is dense.

Allows one to spot an outlier whereby is a lone tick far from others.

```python
# use ecdf graph to indicate how data behave from minimum to maximum
sns.displot(data=df,x='total_bill',kind='ecdf')
```

```python
#put hue=sex
sns.displot(data=df,x='total_bill',kind='ecdf',hue='sex')
```

```python
# add col=day
sns.displot(data=df,x='total_bill',kind='ecdf',hue='sex',col='day')
```

Categorical  Plots

 This kind of plots are useful whe analyzing discrete/categorical data.Example is gender, religion,class.plots like: 

 
 box- show data distribution/spread in quartiles and also outliers.example is employees salary distribution based on departments.

 count plot -which shows number of observations in each category/frequency of category.e.g how many boys vs girls in a class.

 Strip plot- visualises individual data points and adds jitter(jitter is a numeric value like jitter=0.3) jitter aids in avoiding values not to overlap. while swarm plots are similar to strip but the values do not overlap.
 
 Point plots is for trends/comparisons in categories.it shows mean values per category connected by lines.

  N/B in all the plots one has to identify by adding  kind='bar' or the chosen plot.





```python
sns.catplot(data=df,x='day',y='total_bill')
sns.catplot(data=df,x='day',y='total_bill',hue='sex',kind='box')
sns.catplot(data=df,x='day',y='total_bill',kind='violin')
sns.catplot(data=df,x='day',y='total_bill',hue='sex',kind='violin')
sns.catplot(data=df,x='day',y='total_bill',hue='sex',kind='violin',split=True)
sns.violinplot(data=df,x='day',y='total_bill',hue='sex',split=True,inner='quartile')
sns.catplot(data=df,x='day',y='total_bill',kind='bar',hue='sex')
sns.jointplot(data=df,x='total_bill',y='tip',hue='sex',kind='kde')
sns.jointplot(data=df,x='total_bill',y='tip',hue='sex',kind='scatter')
```

```python
# put all together
kind=['scatter', 'kde', 'hist','hex','reg','resid']
for k in kind:
    sns.jointplot(data=df, x='total_bill', y='tip', kind=k)
```
